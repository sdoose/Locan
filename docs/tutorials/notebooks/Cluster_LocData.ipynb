# Tutorial about clustering localizations data

Locan provides methods for clustering localizations in LocData objects. The methods all return a new LocDat object that represents the collected selections for each cluster.

```python
from pathlib import Path

%matplotlib inline

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

import locan as lc
```

```python
lc.show_versions(system=False, dependencies=False, verbose=False)
```

## Synthetic data

We simulate localization data that follows a Neyman-Scott distribution in 2D:

```python
rng = np.random.default_rng(seed=11)
```

```python
locdata = lc.simulate_Thomas(
    parent_intensity=1e-5,
    region=((0, 1000), (0, 1000)),
    cluster_mu=1000,
    cluster_std=10,
    seed=rng,
)

locdata.print_summary()
```

```python
fig, ax = plt.subplots(nrows=1, ncols=1)
locdata.data.plot.scatter(
    x="position_x", y="position_y", ax=ax, color="Blue", label="locdata"
)
plt.show()
```

## Cluster localizations by dbscan

```python
noise, clust = lc.cluster_dbscan(locdata, eps=20, min_samples=3)
assert noise.data.empty
```

```python
fig, ax = plt.subplots(nrows=1, ncols=1)
locdata.data.plot.scatter(
    x="position_x", y="position_y", ax=ax, color="Yellow", label="locdata"
)
lc.LocData.concat(clust.references).data.plot.scatter(
    x="position_x", y="position_y", ax=ax, color="Blue", label="clustered data"
)
clust.data.plot.scatter(
    x="position_x", y="position_y", ax=ax, color="Red", s=10, label="cluster centroids"
)
plt.show()
```

## Cluster localizations in the presence of noise

Often homogeneously distributed localizations are present that cannot be clustered (so-called noise). In this case *noise* should be set True such that two LocData objects are returned that hold noise and the cluster collection. If *noise* is False it will be part of the returned cluster collection.

```python
locdata_cluster = lc.simulate_Thomas(
    parent_intensity=1e-5,
    region=((0, 1000), (0, 1000)),
    cluster_mu=1000,
    cluster_std=10,
    seed=rng,
)
locdata_noise = lc.simulate_Poisson(
    intensity=1e-4, region=((0, 1000), (0, 1000)), seed=rng
)
locdata = lc.LocData.concat([locdata_cluster, locdata_noise])
```

```python
noise, clust = lc.cluster_dbscan(locdata, eps=30, min_samples=3)
```

```python
fig, ax = plt.subplots(nrows=1, ncols=1)
locdata.data.plot.scatter(
    x="position_x", y="position_y", ax=ax, color="Yellow", label="locdata"
)
lc.LocData.concat(clust.references).data.plot.scatter(
    x="position_x", y="position_y", ax=ax, color="Blue", label="clustered data"
)
noise.data.plot.scatter(
    x="position_x", y="position_y", ax=ax, color="Gray", label="noise"
)
clust.data.plot.scatter(
    x="position_x", y="position_y", ax=ax, color="Red", s=50, label="cluster centroids"
)
plt.show()
```

If single localizations should be inlcuded as individual clusters, we need to reduce `min_samples` to 1. In that case `noise` will always be None.

```python
noise, clust = lc.cluster_dbscan(locdata, eps=20, min_samples=1)
assert noise.data.empty
```

```python
fig, ax = plt.subplots(nrows=1, ncols=1)
locdata.data.plot.scatter(
    x="position_x", y="position_y", ax=ax, color="Yellow", label="locdata"
)
lc.LocData.concat(clust.references).data.plot.scatter(
    x="position_x", y="position_y", ax=ax, color="Blue", label="clustered data"
)
# noise.data.plot.scatter(x='position_x', y='position_y', ax=ax, color='Gray', label='noise')
clust.data.plot.scatter(
    x="position_x", y="position_y", ax=ax, color="Red", s=50, label="cluster centroids"
)
plt.show()
```

Other cluster functions are available in the `locan.data.cluster`module.

## Cluster localizations by radius

Cluster localizations by providing fix points and a radius. All localizations within each ball are combined to a cluster.

```python
positions = ((600, 200), (800, 800))
noise, clust, position_indices = lc.cluster_by_radius(
    locdata,
    positions=positions,
    radius=200,
    min_samples=10,
    exclusive=False,
    reduce=False,
    return_noise=True,
)
assert not noise.data.empty
```

```python
fig, ax = plt.subplots(nrows=1, ncols=1)
locdata.data.plot.scatter(
    x="position_x", y="position_y", ax=ax, color="Yellow", label="locdata"
)
ax.scatter(*np.array(positions).T, color="Red", label="positions")
lc.LocData.concat(clust.references).data.plot.scatter(
    x="position_x", y="position_y", ax=ax, color="Blue", label="clustered data"
)
noise.data.plot.scatter(
    x="position_x", y="position_y", ax=ax, color="Gray", label="noise"
)
clust.data.plot.scatter(
    x="position_x",
    y="position_y",
    ax=ax,
    color="Green",
    s=50,
    label="cluster centroids",
)
plt.legend()
plt.show()
```
